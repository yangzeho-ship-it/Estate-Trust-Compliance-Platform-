```python
import pandas as pd
```

```python
df = pd.read_csv("rigel_customer_interviews_realistic_1248.csv")
```

```python
df.columns
```

```python
must_have_df = df[["customer_type","must_haves"]].copy()
must_have_df["must_haves"] = must_have_df["must_haves"].str.split(";")
must_have_df = must_have_df.explode("must_haves")
must_have_df["must_haves"] = must_have_df["must_haves"].str.strip()
must_have_table =pd.crosstab(must_have_df["customer_type"],must_have_df["must_haves"])
must_have_table.to_csv("customer_requirements.csv")
```

```python
customer_requirements_long = (
    must_have_table
    .reset_index()
    .melt(
        id_vars="customer_type",
        var_name="Requirement",
        value_name="Count"
    )
)

customer_requirements_long.to_csv(
    "customer_requirements_long.csv",
    index=False
)
```

```python
product_df = df[['recommended_product_focus']].copy()
product_df["recommended_product_focus"] = product_df["recommended_product_focus"].str.split(".")
product_df = product_df.explode("recommended_product_focus")
product_df["recommended_product_focus"] = product_df["recommended_product_focus"].str.strip()
product_count = product_df['recommended_product_focus'].value_counts().reset_index()
product_count.columns = ['product_focus', 'count']
product_count.to_csv('product_focus_distribution.csv',index=False)
```

```python
product_customer = df[
    ['customer_type','recommended_product_focus']
].copy()

product_customer['recommended_product_focus'] = (
    product_customer['recommended_product_focus']
    .str.split(';')
)

product_customer = product_customer.explode(
    'recommended_product_focus'
)

product_customer['recommended_product_focus'] = (
    product_customer['recommended_product_focus']
    .str.strip()
)
product_customer.to_csv(
    'customer_product_focus.csv',
    index=False
)
```
